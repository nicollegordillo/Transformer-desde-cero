## Bloque 0: Imports, corpus y vocabularios (dado, no modificar)

````python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math, random, hashlib
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

torch.manual_seed(42)
random.seed(42)
print(f'PyTorch: {torch.__version__}')
````

````python
# Corpus SST-2 desde GitHub
import urllib.request, random
urllib.request.urlretrieve(
    'https://raw.githubusercontent.com/clairett/pytorch-sentiment-classification'
    '/master/data/SST2/train.tsv', 'sst2_train.tsv')
urllib.request.urlretrieve(
    'https://raw.githubusercontent.com/clairett/pytorch-sentiment-classification'
    '/master/data/SST2/dev.tsv', 'sst2_dev.tsv')

def filter_data(path, min_len=3, max_len=15):
    out = []
    with open(path) as f:
        for l in f:
            parts = l.strip().split('\t')
            if len(parts) != 2: continue
            text, label = parts[0].lower(), int(parts[1])
            if min_len <= len(text.split()) <= max_len:
                out.append((text, label))
    return out

random.seed(42)
train_all = filter_data('sst2_train.tsv')
dev_all   = filter_data('sst2_dev.tsv')
random.shuffle(train_all)
TRAIN_CLS = train_all[:300]
DEV_CLS   = dev_all[:100]
GPT_TRAIN = [t for t,_ in train_all[:400]]
print(f'Clasificacion - Train: {len(TRAIN_CLS)}, Dev: {len(DEV_CLS)}')
print(f'GPT corpus: {len(GPT_TRAIN)} oraciones')
print(f'Ejemplo clasificacion: {TRAIN_CLS[0]}')
print(f'Ejemplo GPT: {GPT_TRAIN[0]}')
````

````python
# Vocabulario para clasificacion (con token CLS)
PAD, UNK, CLS = '<PAD>', '<UNK>', '<CLS>'
SPECIAL = [PAD, UNK, CLS]
words_cls = Counter()
for t,_ in TRAIN_CLS: words_cls.update(t.split())
vocab_cls = SPECIAL + [w for w,c in words_cls.most_common() if c >= 2]
w2i_cls = {w:i for i,w in enumerate(vocab_cls)}
i2w_cls = {i:w for w,i in w2i_cls.items()}
V_CLS = len(vocab_cls)
MAX_LEN_CLS = 16  # CLS + hasta 15 tokens

# Vocabulario para GPT (con BOS y EOS)
BOS, EOS, PAD_G, UNK_G = '<BOS>', '<EOS>', '<PAD>', '<UNK>'
SPECIAL_G = [PAD_G, UNK_G, BOS, EOS]
words_gpt = Counter()
for t in GPT_TRAIN: words_gpt.update(t.split())
vocab_gpt = SPECIAL_G + [w for w,c in words_gpt.most_common() if c >= 2]
w2i_gpt = {w:i for i,w in enumerate(vocab_gpt)}
i2w_gpt = {i:w for w,i in w2i_gpt.items()}
V_GPT = len(vocab_gpt)
MAX_LEN_GPT = 14  # BOS + hasta 12 tokens + EOS

print(f'Vocab clasificacion: {V_CLS}, MAX_LEN: {MAX_LEN_CLS}')
print(f'Vocab GPT: {V_GPT}, MAX_LEN: {MAX_LEN_GPT}')

def tokenize_cls(text, max_len=MAX_LEN_CLS):
    toks = [CLS] + [w if w in w2i_cls else UNK for w in text.split()[:max_len-1]]
    idxs = [w2i_cls[t] for t in toks]
    mask = [True]*len(idxs) + [False]*(max_len-len(idxs))
    while len(idxs) < max_len: idxs.append(w2i_cls[PAD])
    return torch.tensor(idxs[:max_len]), torch.tensor(mask[:max_len])

def tokenize_gpt(text, max_len=MAX_LEN_GPT):
    toks = [BOS]+[w if w in w2i_gpt else UNK_G for w in text.split()[:max_len-2]]+[EOS]
    idxs = [w2i_gpt[t] for t in toks]
    while len(idxs) < max_len: idxs.append(w2i_gpt[PAD_G])
    return torch.tensor(idxs[:max_len])

# Verificacion de tokenizacion
idxs_test, mask_test = tokenize_cls(TRAIN_CLS[0][0])
print(f'Ejemplo tokenizado: {[i2w_cls[i.item()] for i in idxs_test[:8]]}')
print(f'Mascara: {mask_test[:8].tolist()}')
````

````python
# Dimensiones del modelo (fijas para autograder)
D_MODEL = 32   # dimension central de la arquitectura
D_FF    = 64   # dimension interna del FFN
N_HEADS = 2    # numero de cabezas de atencion
D_K     = D_MODEL // N_HEADS  # dimension por cabeza = 16

# Positional encoding (dado, no modificar)
def make_positional_encoding(max_len, d_model):
    """
    Genera la matriz de positional encoding de Vaswani et al. 2017.
    PE(t, 2i)   = sin(t / 10000^(2i/d_model))
    PE(t, 2i+1) = cos(t / 10000^(2i/d_model))

    Parametros
    ----------
    max_len : int, largo maximo de secuencia
    d_model : int, dimension del modelo

    Retorna
    -------
    pe : torch.Tensor de forma (max_len, d_model)
    """
    pe = torch.zeros(max_len, d_model)
    pos = torch.arange(max_len).unsqueeze(1).float()
    div = torch.exp(
        torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model)
    )
    pe[:, 0::2] = torch.sin(pos * div)
    pe[:, 1::2] = torch.cos(pos * div)
    return pe

PE_CLS = make_positional_encoding(MAX_LEN_CLS, D_MODEL)
PE_GPT = make_positional_encoding(MAX_LEN_GPT, D_MODEL)
print(f'PE_CLS: {PE_CLS.shape}, PE_GPT: {PE_GPT.shape}')

# Mascara causal para GPT (triangular superior = True = enmascarado)
CAUSAL_MASK = torch.triu(torch.ones(MAX_LEN_GPT, MAX_LEN_GPT), diagonal=1).bool()
print(f'CAUSAL_MASK shape: {CAUSAL_MASK.shape}')
print('Primeras 4 filas de la mascara causal:')
print(CAUSAL_MASK[:4, :6].int())
````

## Parte 1: Transformer Encoder para clasificacion de sentimientos

Implemente el encoder Transformer completo como una clase `nn.Module`. El modelo usa un token especial `[CLS]` en la primera posicion: su representacion despues del bloque encoder se usa para clasificar la oracion como positiva (1) o negativa (0).

**Arquitectura:**
```
texto -> tokenize_cls -> E[idxs] + PE  ->  [Multi-Head Self-Attention]
                                            [Add & LayerNorm]
                                            [Feed-Forward por posicion]
                                            [Add & LayerNorm]
                                        ->  representacion CLS -> W_cls -> logits
```

**Parametros que debe declarar en `__init__` como `nn.Parameter`:**

| Parametro | Forma | Descripcion |
|---|---|---|
| `E` | $(V_{cls}, d_{model})$ | Matriz de embeddings |
| `WQ`, `WK`, `WV` | $(d_{model}, d_{model})$ | Proyecciones de atencion |
| `WO` | $(d_{model}, d_{model})$ | Proyeccion de salida multi-head |
| `gamma1`, `beta1` | $(d_{model},)$ | LayerNorm 1 |
| `W1` | $(d_{model}, d_{ff})$ | FFN capa 1 |
| `bias1` | $(d_{ff},)$ | Bias FFN capa 1 |
| `W2` | $(d_{ff}, d_{model})$ | FFN capa 2 |
| `bias2` | $(d_{model},)$ | Bias FFN capa 2 |
| `gamma2`, `beta2` | $(d_{model},)$ | LayerNorm 2 |
| `Wcls` | $(d_{model}, 2)$ | Clasificador sobre token CLS |
| `bcls` | $(2,)$ | Bias clasificador |

````python
def layer_norm(x, gamma, beta, eps=1e-5):
    """
    Layer Normalization sobre la ultima dimension.

    Parametros
    ----------
    x     : Tensor de forma (..., d_model)
    gamma : Tensor de forma (d_model,) — escala aprendible
    beta  : Tensor de forma (d_model,) — desplazamiento aprendible
    eps   : float, estabilidad numerica

    Retorna
    -------
    Tensor de la misma forma que x, normalizado por media y std
    sobre la dimension d_model.
    """
    # =====================
    # SU CODIGO AQUI
    # =====================
    mean = x.mean(-1, keepdim=True)
    var  = x.var(-1, unbiased=False, keepdim=True)
    return gamma * (x - mean) / torch.sqrt(var + eps) + beta

# Prueba rapida de layer_norm
_x = torch.randn(16, D_MODEL)
_g = torch.ones(D_MODEL); _b = torch.zeros(D_MODEL)
_out = layer_norm(_x, _g, _b)
print(f'layer_norm output shape: {_out.shape}')
print(f'Media por fila (debe ser ~0): {_out.mean(-1).abs().max().item():.6f}')
print(f'Std por fila (debe ser ~1):  {_out.std(-1).mean().item():.4f}')
````

````python
class TransformerEncoder(nn.Module):
    """
    Transformer Encoder de un bloque para clasificacion de sentimientos.
    Usa el token [CLS] en la posicion 0 para la clasificacion final.
    """
    def __init__(self, V, d_model, d_ff, n_heads, d_k):
        super().__init__()
        self.d_model = d_model
        self.n_heads = n_heads
        self.d_k = d_k

        # =====================
        # SU CODIGO AQUI
        # Declare todos los parametros como nn.Parameter
        # Inicialice pesos con torch.randn(...) * 0.1
        # Inicialice gamma con ones, beta/biases con zeros
        # =====================
        self.E     = nn.Parameter(torch.randn(V, d_model) * 0.1)
        self.WQ    = nn.Parameter(torch.randn(d_model, d_model) * 0.1)
        self.WK    = nn.Parameter(torch.randn(d_model, d_model) * 0.1)
        self.WV    = nn.Parameter(torch.randn(d_model, d_model) * 0.1)
        self.WO    = nn.Parameter(torch.randn(d_model, d_model) * 0.1)
        self.gamma1 = nn.Parameter(torch.ones(d_model))
        self.beta1  = nn.Parameter(torch.zeros(d_model))
        self.W1    = nn.Parameter(torch.randn(d_model, d_ff) * 0.1)
        self.bias1 = nn.Parameter(torch.zeros(d_ff))
        self.W2    = nn.Parameter(torch.randn(d_ff, d_model) * 0.1)
        self.bias2 = nn.Parameter(torch.zeros(d_model))
        self.gamma2 = nn.Parameter(torch.ones(d_model))
        self.beta2  = nn.Parameter(torch.zeros(d_model))
        self.Wcls  = nn.Parameter(torch.randn(d_model, 2) * 0.1)
        self.bcls  = nn.Parameter(torch.zeros(2))

    def multi_head_attention(self, x, pad_mask):
        """
        Multi-head self-attention sobre la secuencia x.

        Parametros
        ----------
        x        : Tensor (T, d_model) — secuencia con embeddings + PE
        pad_mask : Tensor bool (T,) — True en posiciones validas, False en PAD

        Retorna
        -------
        out      : Tensor (T, d_model)
        attn_w   : Tensor (n_heads, T, T) — pesos de atencion (para visualizacion)

        Pasos:
        1. Proyectar x con WQ, WK, WV: forma (T, d_model)
        2. Reshape a (n_heads, T, d_k) para procesar en paralelo
        3. Calcular scores = Q @ K^T / sqrt(d_k): forma (n_heads, T, T)
        4. Enmascarar posiciones PAD con -inf antes del softmax
        5. Aplicar softmax sobre la ultima dimension
        6. Calcular out = attn_w @ V: forma (n_heads, T, d_k)
        7. Reshape y proyectar con WO: forma (T, d_model)
        """
        T = x.shape[0]
        # =====================
        # SU CODIGO AQUI
        # =====================

        Q = (x @ self.WQ).view(T, self.n_heads, self.d_k).transpose(0, 1)  # (h, T, d_k)
        K = (x @ self.WK).view(T, self.n_heads, self.d_k).transpose(0, 1)
        V = (x @ self.WV).view(T, self.n_heads, self.d_k).transpose(0, 1)

        scores = Q @ K.transpose(-2, -1) / math.sqrt(self.d_k)             # (h, T, T)
        scores = scores.masked_fill(~pad_mask.view(1, 1, T), float('-inf'))

        attn_w = torch.softmax(scores, dim=-1)
        attn_drop = F.dropout(attn_w, p=0.1, training=self.training)   
        out = attn_drop @ V
        out = out.transpose(0, 1).reshape(T, self.d_model) @ self.WO
        return out, attn_w

    def feed_forward(self, x):
        """
        Red feed-forward aplicada por posicion.

        Parametros
        ----------
        x : Tensor (T, d_model)

        Retorna
        -------
        Tensor (T, d_model)

        Formula: FFN(x) = ReLU(x W1 + b1) W2 + b2
        """
        # =====================
        # SU CODIGO AQUI
        # =====================
        h = torch.relu(x @ self.W1 + self.bias1)
        h = F.dropout(h, p=0.1, training=self.training)   
        return h @ self.W2 + self.bias2

    def forward(self, text, return_attn=False):
        """
        Forward pass completo del encoder.

        Parametros
        ----------
        text        : str — oracion de entrada
        return_attn : bool — si True, tambien retorna los pesos de atencion

        Retorna
        -------
        logits : Tensor (2,) — scores para [negativo, positivo]
        attn_w : Tensor (n_heads, T, T) — solo si return_attn=True

        Pasos:
        1. tokenize_cls(text) -> idxs (T,), mask (T,)
        2. Embedding + PE: E[idxs] + PE_CLS -> x (T, d_model)
        3. Multi-head self-attention -> att_out (T, d_model)
        4. Add & LayerNorm: layer_norm(x + att_out, gamma1, beta1)
        5. Feed-forward -> ff_out (T, d_model)
        6. Add & LayerNorm: layer_norm(x2 + ff_out, gamma2, beta2)
        7. Tomar representacion del token CLS: x3[0] (d_model,)
        8. Clasificar: x3[0] @ Wcls + bcls -> logits (2,)
        """
        # =====================
        # SU CODIGO AQUI
        # =====================
        idxs, mask = tokenize_cls(text)
        x  = self.E[idxs] + PE_CLS[:len(idxs)]
        att_out, attn_w = self.multi_head_attention(x, mask)
        x2 = layer_norm(x + att_out, self.gamma1, self.beta1)
        ff_out = self.feed_forward(x2)
        x3 = layer_norm(x2 + ff_out, self.gamma2, self.beta2)
        logits = x3[0] @ self.Wcls + self.bcls
        if return_attn:
            return logits, attn_w
        return logits
````

````python
torch.manual_seed(42)
enc_model = TransformerEncoder(V_CLS, D_MODEL, D_FF, N_HEADS, D_K)

# Test forward en el primer ejemplo
text_test, label_test = TRAIN_CLS[0]
logits_test, attn_test = enc_model(text_test, return_attn=True)

print(f'Texto: "{text_test}"')
print(f'Label: {label_test}')
print(f'logits shape: {logits_test.shape}  (esperado: torch.Size([2]))')
print(f'logits: {logits_test.detach()}')
print(f'attn_w shape: {attn_test.shape}  (esperado: torch.Size([{N_HEADS}, {MAX_LEN_CLS}, {MAX_LEN_CLS}]))')

# Verificar que los attention weights suman 1 por fila
row_sums = attn_test[0].sum(-1)
print(f'Suma por fila head 0 (primeras 4, debe ser ~1.0): {row_sums[:4].detach()}')

# Loss inicial
loss_test = F.cross_entropy(logits_test.unsqueeze(0), torch.tensor([label_test]))
print(f'Loss inicial (esperado ~0.69 = log(2)): {loss_test.item():.4f}')
````

### Entrenamiento del encoder

````python
torch.manual_seed(42)
enc_model = TransformerEncoder(V_CLS, D_MODEL, D_FF, N_HEADS, D_K)
optimizer_enc = torch.optim.Adam(enc_model.parameters(), lr=1e-3)

history_enc = []
for epoch in range(10):
    random.shuffle(TRAIN_CLS)
    total_loss = 0; correct = 0
    enc_model.train()
    for text, label in TRAIN_CLS:
        optimizer_enc.zero_grad()
        logits = enc_model(text)
        loss = F.cross_entropy(logits.unsqueeze(0), torch.tensor([label]))
        loss.backward()
        optimizer_enc.step()
        total_loss += loss.item()
        with torch.no_grad():
            if logits.argmax().item() == label: correct += 1
    avg_loss = total_loss / len(TRAIN_CLS)
    train_acc = correct / len(TRAIN_CLS)
    history_enc.append((avg_loss, train_acc))
    if (epoch+1) % 2 == 0:
        print(f'Epoch {epoch+1:2d}: loss={avg_loss:.4f}, train_acc={train_acc:.2%}')

# Evaluacion en dev
enc_model.eval()
correct_dev = 0
for text, label in DEV_CLS:
    with torch.no_grad():
        logits, _ = enc_model(text, return_attn=True)
        if logits.argmax().item() == label: correct_dev += 1
dev_acc = correct_dev / len(DEV_CLS)
print(f'Dev accuracy: {dev_acc:.2%}')

# Grafica de convergencia
losses = [h[0] for h in history_enc]
accs   = [h[1] for h in history_enc]
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 3))
ax1.plot(losses, 'o-', color='steelblue'); ax1.set_title('Loss por epoca')
ax1.set_xlabel('Epoca'); ax1.set_ylabel('Cross-entropy')
ax2.plot(accs, 'o-', color='darkorange'); ax2.set_title('Accuracy entrenamiento')
ax2.set_xlabel('Epoca'); ax2.set_ylabel('Accuracy')
plt.tight_layout(); plt.savefig('convergencia_encoder.png', dpi=110)
plt.show()
````

## Parte 2: Mini-GPT decoder-only

Implemente un Transformer decoder-only para modelado de lenguaje. El modelo aprende a predecir el siguiente token dado el contexto previo.

**La diferencia clave con el encoder:** la mascara causal. En lugar de que cada token atienda a todos los demas, cada token solo puede atender a los tokens en posiciones anteriores o iguales a la suya.

**Conexion con la Semana 5:** la mascara causal que aplica aqui es exactamente la misma que describimos en el decoder del seq2seq. La diferencia es que aqui no hay cross-attention: el modelo solo tiene self-attention con mascara.

**Reutilize** `layer_norm` y la logica de `multi_head_attention` de la Parte 1. Solo necesita modificar:
1. Agregar la mascara causal en el calculo de scores
2. Cambiar el clasificador CLS por una proyeccion al vocabulario completo
3. Cambiar el objetivo de clasificacion por prediccion del siguiente token

**Parametros del MiniGPT:**

| Parametro | Forma | Descripcion |
|---|---|---|
| `E` | $(V_{gpt}, d_{model})$ | Embeddings GPT |
| `WQ`, `WK`, `WV`, `WO` | $(d_{model}, d_{model})$ | Atencion con mascara causal |
| `gamma1`, `beta1`, `gamma2`, `beta2` | $(d_{model},)$ | LayerNorm |
| `W1`, `bias1`, `W2`, `bias2` | igual que encoder | FFN |
| `Wlm` | $(d_{model}, V_{gpt})$ | Proyeccion al vocabulario (Language Model head) |
| `blm` | $(V_{gpt},)$ | Bias proyeccion |

````python
class MiniGPT(nn.Module):
    """
    Transformer decoder-only para modelado de lenguaje.
    Usa mascara causal: cada token solo atiende a tokens anteriores.
    """
    def __init__(self, V, d_model, d_ff, n_heads, d_k):
        super().__init__()
        self.d_model = d_model
        self.n_heads = n_heads
        self.d_k = d_k
        self.V = V

        # Embeddings
        self.E = nn.Parameter(torch.randn(V, d_model) * 0.1)

        # Parametros de self-attention
        self.WQ = nn.Parameter(torch.randn(d_model, d_model) * 0.1)
        self.WK = nn.Parameter(torch.randn(d_model, d_model) * 0.1)
        self.WV = nn.Parameter(torch.randn(d_model, d_model) * 0.1)
        self.WO = nn.Parameter(torch.randn(d_model, d_model) * 0.1)

        # Primer LayerNorm
        self.gamma1 = nn.Parameter(torch.ones(d_model))
        self.beta1 = nn.Parameter(torch.zeros(d_model))

        # Feed Forward Network
        self.W1 = nn.Parameter(torch.randn(d_model, d_ff) * 0.1)
        self.bias1 = nn.Parameter(torch.zeros(d_ff))
        self.W2 = nn.Parameter(torch.randn(d_ff, d_model) * 0.1)
        self.bias2 = nn.Parameter(torch.zeros(d_model))

        # Segundo LayerNorm
        self.gamma2 = nn.Parameter(torch.ones(d_model))
        self.beta2 = nn.Parameter(torch.zeros(d_model))

        # Language Model Head
        self.Wlm = nn.Parameter(torch.randn(d_model, V) * 0.1)
        self.blm = nn.Parameter(torch.zeros(V))

    def masked_self_attention(self, x):
        """
        Self-attention con mascara causal.
        """
        T = x.shape[0]

        # 1. Proyecciones Q, K y V
        Q = x @ self.WQ
        K = x @ self.WK
        V = x @ self.WV

        # 2. Separar las cabezas
        Q = Q.view(T, self.n_heads, self.d_k).transpose(0, 1)
        K = K.view(T, self.n_heads, self.d_k).transpose(0, 1)
        V = V.view(T, self.n_heads, self.d_k).transpose(0, 1)

        # Formas:
        # Q, K, V -> (n_heads, T, d_k)

        # 3. Scaled dot-product attention
        scores = Q @ K.transpose(-2, -1)
        scores = scores / math.sqrt(self.d_k)

        # 4. Mascara causal:
        # posiciones futuras reciben -infinito
        causal_mask = CAUSAL_MASK[:T, :T]
        scores = scores.masked_fill(
            causal_mask.unsqueeze(0),
            float('-inf')
        )

        # 5. Softmax
        attn_w = torch.softmax(scores, dim=-1)

        # 6. Aplicar atencion sobre V
        out = attn_w @ V

        # (n_heads, T, d_k) -> (T, d_model)
        out = out.transpose(0, 1).reshape(T, self.d_model)

        # 7. Proyeccion final
        out = out @ self.WO

        return out, attn_w

    def feed_forward(self, x):
        """
        FFN(x) = ReLU(x W1 + b1) W2 + b2
        """
        hidden = torch.relu(x @ self.W1 + self.bias1)
        return hidden @ self.W2 + self.bias2

    def forward_all(self, idxs):
        """
        Forward sobre una secuencia completa de indices.
        """
        T = len(idxs)

        # 1. Embedding + positional encoding
        x = self.E[idxs] + PE_GPT[:T]

        # 2. Masked self-attention
        att_out, attn_w = self.masked_self_attention(x)

        # 3. Residual + LayerNorm
        x2 = layer_norm(
            x + att_out,
            self.gamma1,
            self.beta1
        )

        # 4. Feed-forward
        ff_out = self.feed_forward(x2)

        # 5. Residual + LayerNorm
        x3 = layer_norm(
            x2 + ff_out,
            self.gamma2,
            self.beta2
        )

        # 6. Proyeccion al vocabulario
        logits = x3 @ self.Wlm + self.blm

        return logits, attn_w

    def forward(self, text):
        """
        Calcula la loss de language modeling.
        """
        idxs = tokenize_gpt(text)

        logits, _ = self.forward_all(idxs)

        pad_idx = w2i_gpt[PAD_G]

        # Cada posicion predice el siguiente token:
        #
        # logits[0] -> idxs[1]
        # logits[1] -> idxs[2]
        # ...
        loss = F.cross_entropy(
            logits[:-1],
            idxs[1:],
            ignore_index=pad_idx
        )

        return loss

    @torch.no_grad()
    def generate(self, seed_text, max_new=8, temperature=0.9):
        """
        Generacion autoregresiva.
        """
        self.eval()

        words = seed_text.split()

        for _ in range(max_new):

            # Texto generado hasta ahora
            cur = ' '.join(words)

            # Tokenizar
            idxs = tokenize_gpt(cur)

            # Forward
            logits, _ = self.forward_all(idxs)

            # Posicion del ultimo token valido
            n_valid = min(
                len(words) + 1,
                MAX_LEN_GPT - 1
            )

            next_logits = logits[n_valid - 1] / temperature

            # No permitir generar PAD ni BOS
            next_logits[w2i_gpt[PAD_G]] = float('-inf')
            next_logits[w2i_gpt[BOS]] = float('-inf')

            # Convertir logits a probabilidades
            probs = torch.softmax(next_logits, dim=-1)

            # Muestrear siguiente token
            next_idx = torch.multinomial(
                probs,
                num_samples=1
            ).item()

            next_word = i2w_gpt[next_idx]

            # Detener si genera EOS o PAD
            if next_word == EOS or next_word == PAD_G:
                break

            # Agregar palabra generada
            words.append(next_word)

        return ' '.join(words)
````

**Su respuesta a la Pregunta 1:**

**a) Flujo de información desde 'surprisingly' hasta `[CLS]`**

Sea $j$ la posición del token `surprisingly` y $0$ la del `[CLS]`. La información viaja
en cinco etapas, cada una con una operación matricial identificable.

**1. Entrada.** El token se convierte en vector por indexación de la matriz de
embeddings, más su codificación posicional:

$$x_j = E[\text{idx}_j] + PE[j] \in \mathbb{R}^{32}$$

$E$ aporta qué palabra es y $PE$ aporta dónde está. En este punto $x_j$ todavía no
sabe nada del resto de la oración.

**2. Proyecciones.** Cada posición se proyecta a tres espacios distintos:

$$q_0 = x_0 W_Q, \qquad k_j = x_j W_K, \qquad v_j = x_j W_V$$

$W_Q$ codifica qué busca el `[CLS]`, $W_K$ codifica qué ofrece `surprisingly` como candidato a ser atendido, y $W_V$ codifica qué contenido transmite si resulta atendido. Que sean tres matrices independientes es lo que permite que "ser relevante para una consulta" y "el contenido que se transmite" sean funciones separadas del mismo vector.

**3. Score y normalización.**

$$s_{0j} = \frac{q_0 \cdot k_j}{\sqrt{d_k}} = \frac{q_0 k_j^\top}{\sqrt{16}},
\qquad
\alpha_{0j} = \frac{\exp(s_{0j})}{\sum_{j'} \exp(s_{0j'})}$$

El $\sqrt{d_k}$ evita que los scores escalen con $d_k$ y saturen el softmax en una
distribución casi one-hot, lo que anularía los gradientes. El $\alpha_{0j}$ resultante es exactamente el número que aparece en la fila 0 del heatmap.

**4. Agregación.**

$$\text{head}_h(0) = \sum_{j'} \alpha_{0j'} v_{j'}$$

Este es el único punto de todo el bloque donde información de posiciones distintas se mezcla. El FFN y el LayerNorm operan posición por posición; la atención es el único canal de comunicación entre tokens.

**5. Salida.** Las dos cabezas se concatenan y se proyectan con $W_O$, lo que permite que se combinen en vez de quedar en subespacios aislados. Después:

$$x^{(2)}_0 = \text{LN}(x_0 + \text{attn}_0), \quad
x^{(3)}_0 = \text{LN}(x^{(2)}_0 + \text{FFN}(x^{(2)}_0)), \quad
\text{logits} = x^{(3)}_0 W_{cls} + b_{cls}$$

El camino completo es
$E[\text{surprisingly}] \to v_j \to \alpha_{0j}v_j \to W_O \to$ residual $\to$ LN $\to$
FFN $\to W_{cls}$. Y hay un segundo camino: el embedding también determina
$k_j$, que determina $\alpha_{0j}$, que pondera a todos los values. O sea que
`surprisingly` influye tanto en el contenido que aporta como en cuánto se le hace caso a los demás tokens.

**c) Qué captura una segunda capa que la primera no puede**

La diferencia está en sobre qué se calculan los scores de atención.

En la primera capa los scores dependen solo de embeddings estáticos:

$$s^{(1)}_{ij} = \frac{(E[\text{idx}_i]+PE[i])\, W_Q W_K^\top\, (E[\text{idx}_j]+PE[j])^\top}{\sqrt{d_k}}$$

Como $W_Q W_K^\top$ es una matriz fija, esto es una forma bilineal sobre pares de
tokens: la afinidad entre las posiciones $i$ y $j$ queda completamente determinada por qué palabras son y dónde están. La primera capa no puede modular su atención según el contexto, porque al momento de calcular sus scores el contexto todavía no existe.

En la segunda capa la entrada ya es $x^{(1)}$, donde cada $x^{(1)}_i$ es una mezcla
ponderada de toda la oración:

$$s^{(2)}_{ij} \propto x^{(1)}_i W_Q W_K^\top {x^{(1)}_j}^\top
= \Big(\textstyle\sum_p \alpha_{ip} v_p\Big) W_Q W_K^\top \Big(\sum_q \alpha_{jq} v_q\Big)^\top$$

Expandiendo, es una suma sobre pares $(p,q)$: la afinidad entre $i$ y $j$ depende de
interacciones entre cuatro tokens, no dos. Eso habilita dos cosas inalcanzables con
una sola capa:

1. **Atención condicionada al contexto.** `bank` tiene un solo embedding, así que en la capa 1 su afinidad con cualquier token es fija. En la capa 2 su representación ya absorbió `river` o `money` y puede atender a cosas distintas según el sentido. La desambiguación léxica requiere mínimo dos capas.

2. **Composición jerárquica.** En `not very good`, la capa 1 puede ligar `not` con
   `very` y `very` con `good`. La capa 2 recibe en la posición de `very` un vector que ya contiene ambos, y puede tratar `not very good` como una unidad y relacionarla con el `[CLS]`. Con una sola capa la negación de una frase completa es imposible, porque `not` solo alcanza a modificar tokens individuales, nunca constituyentes.

Es la misma intuición que en visión: una capa detecta bordes, la siguiente detecta
esquinas hechas de bordes. BERT usa 12 capas porque la sintaxis natural es recursiva y cada nivel de recursión necesita un nivel de composición.

**Su respuesta a la Pregunta 2:**

**a) Por qué $\exp(-\infty)=0$ y qué pasaría con $-10^6$**

En IEEE 754, `float('-inf')` cumple $\exp(-\infty)=0$ exactamente, no
aproximadamente. Como

$$\alpha_{ij} = \frac{\exp(s_{ij})}{\sum_{j'}\exp(s_{ij'})}$$

un numerador exactamente cero da $\alpha_{ij}=0$ exactamente. Esto importa por dos razones:
la posición futura no aporta nada a $\sum_j\alpha_{ij}v_j$, y su gradiente también es cero, así que no hay fuga de información ni por el forward ni por el backward. En la verificación de la Parte 2A esto se ve en que el máximo del triángulo superior salió `0.000000` y no `1e-9`.

Con $-10^6$ el resultado numérico es casi siempre idéntico, y explicar por qué es lo
interesante. PyTorch implementa softmax de forma estable restando el máximo de la fila:

$$\alpha_{ij} = \frac{\exp(s_{ij}-\max_k s_{ik})}{\sum_{j'}\exp(s_{ij'}-\max_k s_{ik})}$$

Con scores reales del orden de $\pm 10$, el exponente enmascarado queda en
$\approx -10^6$, y $\exp(-10^6)$ hace underflow a 0 en float32 (ya $\exp(-104)$ cae
por debajo del mínimo subnormal, del orden de $10^{-45}$). Pero hay tres diferencias que sí importan:

1. **En float16 se rompe.** El máximo representable en half precision es 65504, así que $-10^6$ no es representable y produce `inf` o `NaN` al hacer el cast. Por eso las implementaciones reales usan `torch.finfo(dtype).min` y no una constante fija.

2. **$-\infty$ falla ruidosamente; $-10^6$ falla en silencio.** Si una fila queda
   completamente enmascarada, con $-\infty$ obtienes $0/0=$ `NaN` y te enteras de
   inmediato; con $-10^6$ obtienes una distribución uniforme sobre posiciones inválidas, un bug que contamina el entrenamiento sin que nada se queje.

3. **Los gradientes.** Con $-\infty$ el gradiente respecto al score enmascarado es
   idénticamente cero por construcción. Con $-10^6$ es cero por underflow, pero pasa por una multiplicación que puede producir `NaN` si el otro factor es `inf`.

*(En nuestro encoder aplicamos `masked_fill` con $-\infty$ solo sobre columnas y nunca sobre filas completas, precisamente para evitar el caso 2.)*

**b) Gradiente de la posición 0 vs. la posición $T-1$**

La loss es la cross-entropy promediada sobre las posiciones:

$$\mathcal{L} = \frac{1}{T-1}\sum_{t=0}^{T-2} \text{CE}(\text{logits}_t,\ \text{idx}_{t+1})$$

Con máscara causal, el embedding de la posición $t$ influye en $\text{logits}_s$ solo para $s \geq t$. Por lo tanto:

- Posición 0: visible desde las $T$ posiciones, recibe contribuciones de los $T-1$ términos de la suma:
  $$\frac{\partial\mathcal{L}}{\partial x_0} = \sum_{s=0}^{T-2}\frac{\partial\mathcal{L}_s}{\partial x_0}$$
- Posición $T-1$: solo se ve a sí misma, y su predicción corresponde al token $T$
  que no existe o es PAD (descartado por `ignore_index`). Recibe cero o casi cero.

La consecuencia es una asimetría sistemática: las posiciones tempranas acumulan del
orden de $T$ veces más señal y se entrenan más rápido. Con $T=14$, la posición 0 recibe hasta 13 contribuciones y la última esencialmente ninguna.

Se compone con un efecto en el mismo sentido: el token `<BOS>` está en la posición 0 de todas las oraciones, así que su embedding acumula gradiente de las 400 oraciones × 13 posiciones, mientras que una palabra rara aparece dos veces en total. El resultado neto es un modelo que predice bien el segundo token y cada vez peor conforme avanza.

**c) Temperatura y forma de la distribución**

La temperatura reescala los logits antes del softmax:

$$p_i(\tau) = \frac{\exp(z_i/\tau)}{\sum_j \exp(z_j/\tau)}$$

El efecto es sobre las **razones** entre probabilidades:

$$\frac{p_i(\tau)}{p_j(\tau)} = \exp\!\left(\frac{z_i-z_j}{\tau}\right)
= \left[\frac{p_i(1)}{p_j(1)}\right]^{1/\tau}$$

O sea que la temperatura eleva la distribución original a la potencia $1/\tau$ y
renormaliza. De ahí los dos límites:

- **$\tau\to 0$:** el exponente $1/\tau\to\infty$, cualquier ventaja se amplifica hasta dominar, y la distribución converge a one-hot sobre $\arg\max_i z_i$ (muestreo
greedy determinista). A $\tau=0.5$ ya estamos cerca: una brecha de logits de 1.0 pasa de una razón 2.7:1 a una de 7.4:1.
- **$\tau\to\infty$:** el exponente $\to 0$, todas las razones tienden a 1 y la
distribución converge a uniforme sobre las 432 palabras, o sea ruido.

Formalmente, la entropía $H(p(\tau)) = -\sum_i p_i\log p_i$ crece monótonamente con
$\tau$, y la entropía es literalmente la medida de cuánta variedad hay en el muestreo.

**Resultados observados.**

| $\tau=0.5$ | $\tau=1.5$ |
|---|---|
| `the film is a <UNK> <UNK> of <UNK>` | `the film a ending going , but the moving showgirls` |
| `the story 's <UNK>` | `the story looks tale that for a triumph` |
| `not not an <UNK>` | `not empty will feel his dull an years we` |
| `a surprisingly <UNK>` | `a surprisingly 's clever engrossing emotional inconsequential , it character` |

Lo esperado se cumple en lo estructural: a $\tau=0.5$ las salidas son cortas, repetidas entre corridas y con bucles evidentes (`not not an`); a $\tau=1.5$ son largas, variadas y sintácticamente desarmadas.

Un resultado contraintuitivo que se pudo observar. Midiendo sobre 120
generaciones por temperatura, la fracción de tokens generados que son `<UNK>` es:

- $\tau=0.5$: **62.1%**
- $\tau=1.5$: **8.9%**

O sea que la temperatura baja produce más `<UNK>`, no menos. La explicación se
sigue directamente de la fórmula: `<UNK>` representa el 28.5% del corpus de
entrenamiento, así que es el token individualmente más probable en la mayoría de los
contextos, y por tanto es el argmax. Bajar la temperatura colapsa sobre el argmax, y
el argmax es `<UNK>`. Subirla reparte masa hacia la cola, donde viven las palabras
reales, que individualmente son improbables pero colectivamente cargan la mayor parte de la probabilidad.

Esto refina la lectura habitual de "temperatura baja = mejor calidad": lo que la
temperatura baja maximiza es la probabilidad bajo el modelo, y si el modelo está mal
calibrado (porque el corpus es diminuto) el modo de la distribución no es la salida más útil. También explica la diferencia de longitud: a $\tau=0.5$ el promedio es de 3.5 tokens nuevos contra 6.4 a $\tau=1.5$, porque `<EOS>` también es de alta probabilidad y el muestreo greedy lo elige temprano.

**Su respuesta a la Pregunta 3:**

**a) De self-attention a cross-attention**

En `masked_self_attention` las tres proyecciones parten del mismo tensor `x`:

```python
Q = (x @ self.WQ).view(T, self.n_heads, self.d_k).transpose(0, 1)
K = (x @ self.WK).view(T, self.n_heads, self.d_k).transpose(0, 1)
V = (x @ self.WV).view(T, self.n_heads, self.d_k).transpose(0, 1)
scores = Q @ K.transpose(-2, -1) / math.sqrt(self.d_k)
scores = scores.masked_fill(CAUSAL_MASK[:T, :T], float('-inf'))
```

Para convertirlo en cross-attention hay que cambiar tres cosas:

```python
def cross_attention(self, x, enc_out):                    # (1) nuevo argumento
    T_dec, T_enc = x.shape[0], enc_out.shape[0]
    Q = (x @ self.WQ).view(T_dec, self.n_heads, self.d_k).transpose(0, 1)
    K = (enc_out @ self.WK).view(T_enc, self.n_heads, self.d_k).transpose(0, 1)   # (2)
    V = (enc_out @ self.WV).view(T_enc, self.n_heads, self.d_k).transpose(0, 1)   # (2)
    scores = Q @ K.transpose(-2, -1) / math.sqrt(self.d_k)   # ahora (h, T_dec, T_enc)
    # (3) se elimina CAUSAL_MASK; en su lugar, mascara de padding del encoder
    attn_w = torch.softmax(scores, dim=-1)
    out = (attn_w @ V).transpose(0, 1).reshape(T_dec, self.d_model) @ self.WO
    return out, attn_w
```

1. Recibir la salida del encoder como segundo argumento.
2. Proyectar $K$ y $V$ desde `enc_out` en vez de `x`. $Q$ se queda en el decoder.
3. Quitar la máscara causal.

Estas bastan porque el mecanismo de atención nunca supuso que $Q$, $K$ y $V$ vinieran
de la misma secuencia: la fórmula $\text{softmax}(QK^\top/\sqrt{d_k})V$ solo requiere que $Q$ y $K$ vivan en el mismo espacio de dimensión $d_k$ para que el producto punto tenga sentido, y que $K$ y $V$ tengan la misma longitud para que la ponderación sea válida.
Self-attention es el caso particular en que ambas secuencias coinciden; la matriz de
scores deja de ser cuadrada y pasa a ser $T_{dec}\times T_{enc}$, pero nada más cambia.

El tercer cambio se sigue del segundo: la máscara causal existe para impedir que la
posición $t$ vea el futuro de su propia secuencia, porque en entrenamiento el target
está en el input desplazado. Pero la secuencia fuente está completa y disponible desde el inicio — traducir la tercera palabra legítimamente requiere ver la oración fuente entera. Enmascarar aquí no sería solo innecesario, sería incorrecto.

Con eso, un bloque de decoder del Transformer original queda como: self-attention
enmascarada (lo ya implementado) → cross-attention (esto) → FFN, cada uno con su
residual y su LayerNorm.

**c) Qué no puede aprender el Mini-GPT por limitaciones del corpus**

Los números del corpus son contundentes:

- 400 oraciones, **3,976 tokens en total**. GPT-3 vio 300 mil millones: 75 millones de veces más.
- 1,561 tipos distintos, de los cuales 1,133 aparecen una sola vez (72.6% son
hapax legomena).
- El umbral `count >= 2` deja el vocabulario en 432 palabras, convirtiendo el 28.5% de todos los tokens del corpus en `<UNK>`.
- 3,062 bigramas distintos, de los cuales solo 229 aparecen más de una vez.

De ahí salen cuatro capacidades inalcanzables:

**1. Semántica léxica más allá de las palabras funcionales.** Casi un tercio del corpus es `<UNK>`, que termina siendo uno de los tokens más frecuentes y por tanto una continuación estadísticamente probable en casi cualquier contexto. El efecto medido es brutal: a temperatura 0.5, el 62% de los tokens generados son `<UNK>`. El modelo no aprendió palabras, aprendió que "algo desconocido va aquí".

**2. Sintaxis de largo alcance.** Con 229 bigramas repetidos, el modelo memoriza
colocaciones de dos o tres palabras de alta frecuencia (`the` aparece 172 veces, `a`
162, `of` 98). No hay evidencia suficiente para inducir concordancia sujeto-verbo a
distancia, estructura de subordinadas ni dependencias que crucen más de tres tokens.
Además $\text{MAX\_LEN\_GPT}=14$ impone un techo arquitectónico: el modelo nunca observa
una dependencia más larga, así que no podría aprenderla aunque el corpus la contuviera.

**3. Conocimiento factual y de mundo.** No hay nada que extraer de 4,000 tokens de
reseñas. El modelo no sabe qué es una película; solo que ciertas secuencias de símbolos siguen a otras.

**4. Coherencia de discurso.** Cada ejemplo es una oración aislada delimitada por
`<BOS>`/`<EOS>`. El modelo nunca observa dos oraciones consecutivas, así que la noción de que un texto continúa a otro no existe en su distribución. Por eso `<EOS>` es de alta probabilidad y las generaciones a temperatura baja se cortan en 3–4 tokens.

**Cómo se manifiesta.** El patrón general es **plausibilidad local con incoherencia
global**. `the story looks tale that for a triumph` tiene bigramas defendibles
(`the story`, `a triumph`) pero se desarma como oración. `not not an <UNK>` muestra el bucle repetitivo del muestreo greedy. Y `a surprisingly 's clever engrossing emotional inconsequential , it character` acumula adjetivos de reseña sin ninguna estructura que los organice o sea, el modelo aprendió qué palabras habitan este dominio, no cómo combinarlas.

El punto de fondo, que conecta con el resto del curso: el cuello de botella no es la arquitectura. Los mecanismos implementados (atención multi-cabeza, máscara causal,
residuales, LayerNorm) son exactamente los de GPT-3. Lo que falta son datos y escala: la misma arquitectura corriendo con seis órdenes de magnitud menos de todo.
